```
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
```

```
A = {1,2,3}
B = {2,3,4}

R = {(1,2),(2,3),(3,2),(3,4)}

print("관계 R: ",R)
```

```
A_list = sorted(A)
B_sorted = sorted(B)

relation_matrix = np.zeros((len(A),len(B)), dtype=int)

for i, a in enumerate(A_list):
    for j, b in enumerate(B_sorted):
        if (a,b) in R:
            relation_matrix[i,j] = 1

print("관계 행렬 (rows: A, columns: B): ")
print(relation_matrix)
```

```
print("R =", R)
print("R의 자료형 =", type(R))
print("R의 원소 =", list(R))
```

```
C = {'x','y'}
S = {(2,'x'),(3,'y'),(4,'x')}

composed_relation = set()

for a, b1 in R:
    for b2, c in S:
        if b1 == b2:
            composed_relation.add((a,c))

print(composed_relation)
```
