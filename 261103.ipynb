```python
graph = {
    'Alice' : ['Bob', 'Charlie'],
    'Bob' : ['David','Eve'],
    'Charlie' : ['Frank'],
    'David' : [],
    'Eve' : ['Frank'],
    'Frank' : []
}
```

```python
from collections import deque
```

```python
def dfs(graph,v,visited):
  visited.add(v)
  print(v,end=' ')
  for neighbor in graph[v]:
    if neighbor not in visited:
      dfs(graph,neighbor,visited)
```

```python
print("DFS 탐색 순서:")
dfs(graph,'Alice',set())
```

```python
def bfs(graph,start):
  visited = set()
  queue = deque([start])
  visited.add(start)

  while queue:
    v = queue.popleft()
    print(v, end=' ')
    for neighbor in graph[v]:
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)
```

```python
print("\nBFS 탐색 순서:")
bfs(graph,'Alice')
```

```python
def recommend_friend(graph,start):
  visited = set()
  queue = deque([(start,0)])

  print(f"\n{start}의 친구 추천")

  while queue:
    current, depth = queue.popleft()
    if current not in visited:
      visited.add(current)
      if depth == 2:
        print("-",current)
      if depth < 2:
        for neighbor in graph[current]:
          queue.append((neighbor,depth+1))
```

```python
recommend_friend(graph,'Alice')
```
